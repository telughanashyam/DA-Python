```python
# LOWER CASE , UPPER CASE , CAPITALIZE , TITLE , SWAP CASE , CASEFOLD

a= "ghana shyam"
print(a.upper())
print(a.lower())
print(a.capitalize())
print(a.title())
print(a.swapcase())
print(a.casefold())
```

```python
# STRIP = IT WILL REMOVE THE SPACES IN THE LEFT OF RIGHT
# IF WE WANT TO REMOVE ONLY LEFT SPACE = LSTRIP
# IF WE WANT TO REMOVE ONLY RIGHT SPACE = RSTRIP
a= "        a           b       c    "
print(a.lstrip())
print(a.rstrip())
print(a.strip())
```

```python
a= "***ghana shyam***"
print(a.strip("*"))
print(a.strip("a"))
```
