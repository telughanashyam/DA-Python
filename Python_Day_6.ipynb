```python
# find the second highest

a=int(input())
b=int(input())
c=int(input())
if (a>b and a<c) or (a<b and a>c):
  print("a is second highest")
elif (b>a and b<c) or (b<a and b>c):
  print("b is second highest")
elif (c>a and c<b) or (c<a and a>b):
  print("c is second highest")
else:
  print("a,b,c are equal")
```

```python
# check weather the year is leap year or not

x=int(input())
if (x%4==0) and (x%400==0):
  print("leap year",x)
elif x%100!=0:
  print("leap year",x)
else:
  print("not a leap year",x)
```

```python
# if the student got less than 35 in any subject he shuold get fail
# first line show total marks
# second line show avg marks
# third line show grade

maths,science,english=map(int,input().split())
total=maths+science+english
print(total)
avg=((maths+science+english)/3)
print(avg)
if maths<35 or science<35 or english<35:
  print("Grade f")
elif avg>=90 and avg==100:
  print("Grade = A")
elif avg>75 and avg<89:
  print("Grade = B")
elif avg>60 and avg<74:
  print("Grade = c")
elif avg>40 and avg<59:
  print("Grade = D")
else:
  print("Grade = Fail")

```
