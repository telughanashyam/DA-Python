```python
# LOOPS is used to run same task multiple times
# two types of LOOPS 1] FOR LOOP 2] WHILE LOOP

# FOR LOOP is a numerical loop
# if we known starting point and ending point we can use FOR LOOP

# WHILE LOOP or UNconditional loop
# if we known starting point and dont kmown the ending point we can use WHILE LOOP
# until the iterative/conditonal statement fails
```

```python
# Print 1-10
for i in range(0,11,1):
  print(i)
```

```python
# EVEN Numbers
for i in range(0,11,2):
  print(i)
```

```python
# ODD numbers
for i in range(1,10,2):
  print(i)
```

```python
# Decriment even
for i in range(10,1,-2):
  print(i)
```

```python
# Decriment Odd
for i in range(9,0,-2):
  print(i)
```

```python
# Multiplication table
n=int(input())
for i in range(1,11):
  print(f"{n}x{i}={n*i}")
```

```python
# for loop String
s=input()
for i in s:
  print(i)
```

```python
# Count the number of letters in the string

s=input()
c=0
for i in s:
  c=c+1
print(c)

```

```python
# Count total numbers
l=int(input())
u=int(input())
sum=0
for i in range(l,u+1):
  sum=sum+i
print(sum)
```

```python
# Count only vowels
s=input()
c=0
for i in s:
  if i in "a,e,i,o,u,A,E,I,O,U":
    c=c+1
print(c)
```

```python
# Calculator and ask the operator 5 times

a=int(input())
b=int(input())
for i in range(1,6):
  o=input()
  if o=="+":
    print(a+b)
  elif o=="-":
    print(a-b)
  elif o=="*":
    print(a*b)
  elif o=="%":
    print(a%b)
  elif o=="/":
    print(a/b)
  elif o=="//":
    print(a//b)
  elif o=="**":
    print(a**b)
  else:
    print("invalid operator")
```

```python
# print the sum of 1 to n numbers
n=int(input())
sum=0
for i in range(1,n+1):
  sum=sum+i
print(sum)
```

```python
# print the numbers that are divisible by both 3 and 5 in between (1 to 100)
for i in range(1,101):
  if i%3==0 and i%5==0:
    print(i)
```

```python
# print the squares and cubes of 1 to n numbers
n=int(input())
for i in range(1,n+1):
  print(i, i**2, i**3)
```
