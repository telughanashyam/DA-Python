BITWISE OPERATORS
------------------
&   AND          1 if both bits are 1

`   OR

^   XOR          1 if bits are different

~   NOT          invert's bits

<<  LEFT SHIFT   shifts to left

>>  RIGHT SHIFT  shifts to right

```python
a=5
b=3
print(a&b)
```

```python
a=7
b=3
print(a&b)
```

```python
x=9
y=2
z=3
print(x&y&z)
```

```python
x=4
y=9
print(x|y)
```
