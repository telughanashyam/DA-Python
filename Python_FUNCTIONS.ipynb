User-Define Functions

```python
# creating a function

def greet(name): # greet is a function_name and "name" is parameters
  print("hello",name)
# calling a function
greet("ghana") # name is parameter and ghana is argument
# return must and should write inside the functions


```

```python
# No parameters and No returns
def welcome():
  print("welcome to world")
welcome()
```

```python
# with parameter and no return
def welcome(name):
  print("welcome to world ",name)
welcome("ghana shyam")
```

```python
# No parameters and With return
def welcome():
  return "welcome to codegan"
welcome()
```

```python
# with parameter and with return
def welcome(name):
  return("welcome to codegan",name)
welcome("ghana")
```

```python
# Create a function that prints the multiplication table of 5 from 1 to 10.
def table():
  a=int(input())
  i=1
  for i in range(1,11):
    t=a*i
    print(a,"*",i,"=",t)
    i=i+1
table()
```

```python
# Create a function that accepts two numbers and prints their sum, difference, product, and division.
def operations():
  a=int(input())
  b=int(input())
  print("Add = ",a+b)
  print("Diff = ",a-b)
  print("Product = ",a*b)
  print("Div = ",a/b)
operations()
```
