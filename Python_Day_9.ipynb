```python
# PRIME NUMBER

n=int(input())
if n>=2:
  for i in range(2,n):
    if n%i==0:
      print("Not a Prime")
      break
  else:
    print("Prime")
```

```python
# Numbers are Perfect Number are Not a Perfect Number

n=int(input())
sum=0
for i in range(1,n):
  if n%i==0:
    sum=sum+i
if n==sum:
  print("Perfect Number")
else:
  print("Not a Perfect Number")
```

```python
# To PRINT the FACTORS OF Numbers that are Perfect Number or Not a Perfect Number
n=int(input())
sum=0
for i in range(1,n):
  if n%i==0:
    sum=sum+i
    print(i)
if n==sum:
  print("Perfect Number")
else:
  print("Not a Perfect Number")
```

```python
# Factors

n=int(input())
for i in range(1,n+1):
    if n%i==0:
        print(i,end=' ')
```
