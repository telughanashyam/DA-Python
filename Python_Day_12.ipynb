```python
# guess the correct number using while loop

num=int(input())
while True:
  guess=int(input())
  if guess==num:
    print("correct")
    break
  elif guess < num:
    print("guess is lesser then num")
  elif guess > num:
    print("guess is greater then num")
  else:
    print("not matched")
```

```python
# Guess the number between 1 to 10

num=int(input())
while True:
  if num > 10:
    print("invalid guess")
    break
  guess=int(input())
  if guess==num:
    print("correct")
    break
  elif guess < num:
    print("guess is lesser then num")
  elif guess > num:
    print("guess is greater then num")
  else:
    print("not matched")
```

```python
# ATM pin verfication using while loop

Pin = int(input("should be 4 digits = "))
while True:
  if Pin <= 999 or Pin > 9999:
    print("invalid")
    break
  Num=int(input())
  if Pin==Num:
    print("correct Pin")
    break
  else:
    print("incorrect")
```
