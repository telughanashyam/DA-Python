DICTIONARY

```python
# create a dictinoary using {}

student = {"name":"Ghana","age":"22","marks":"100"}
print(type(student))
print(student)
print(student["name"])
print(student["age"])
print(student["marks"])
```

```python
# adding element in the dictionary

student = {"name":"Ghana","age":"22","marks":"100"}
student["course"]="Python"
student["marks"]="60"
print(student)
```

```python
# updating elements

student = {"name":"Ghana","age":"22","marks":"100","course":"python"}
student["name"]= "shyam"
student["age"]="23"
student["marks"]="100"
print(student)
```

```python
# create a dictionary of employe details with five details

employee= {"name":"sai","age":"25","role":"Data analytics","salary":"50000","gender":"male"}
print(employee)
```

```python
# update employee salary 1,25,000
employee= {"name":"sai","age":"25","role":"Data analytics","salary":"50000","gender":"male"}
employee["salary"]="125000"
print(employee)
```

```python
# add new employe details as blood group b+
employee= {"name":"sai","age":"25","role":"Data analytics","salary":"50000","gender":"male"}
employee["blood group"]="B+"
print(employee)
```

```python
# Removeing dictionary values

employee= {"name":"sai","age":"25","role":"Data analytics","salary":"50000","gender":"male","Blood group":"B+"}
employee.clear()
print(employee)
```

```python
# for printing keys

employee= {"name":"sai","age":"25","role":"Data analytics","salary":"50000","gender":"male"}
for i in employee.keys():
  print(i)
```

```python
# for printing values
employee= {"name":"sai","age":"25","role":"Data analytics","salary":"50000","gender":"male"}
for i in employee.values():
  print(i)
```

```python
# printing both keys and values

for i,j in employee.items():
  print(f'{i}:{j}')
```

```python
# Get()
print(employee.get("role"))
```

```python
dis=int(input())
if dis<=2:
  fare=50
elif dis<10:
  fare=50+(dis-2)*15
else:
  fare=170+(dis-10)*10
if dis>20:
  total_fare=fare*0.05
else:
  total_fare=fare
print("total fare ",total_fare)
```
