```python
# Take a number n as input and find the sum of numbers from 1 to n

n=int(input())
i=1
t=0
while i<=n:
  t=i+t
  i=i+1
print(t)
```

```python
# take a number as input and find sum of all the even numbers

n=int(input())
i=2
t=0
while i<=n:
  t=i+t
  i=i+2
print(t)
```

```python
# take a number as input and find sum of all the odd numbers

n=int(input())
i=1
t=0
while i<=n:
  t=t+i
  i=i+2
print(t)
```

```python
# take a number as input and print sum of digits (even numbers)

num=int(input())
sum=0
n=str(num)
i=0
while i <len(n):
  if int(n[i])%2==0:
    sum=sum+int(n[i])
  i=i+1
print(sum)
```
